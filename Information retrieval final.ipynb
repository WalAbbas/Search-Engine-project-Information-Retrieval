```python
import pandas as pd
import numpy as np
from query_processor import preprocess_query
from retrieval_models import RetrievalSystem

# 
# SECTION 1 — QUERY SET (20 Evaluation Queries)
# 

TEST_QUERIES = {
    "Keyword": [
        "blockchain consensus algorithms",
        "quantum entanglement computing",
        "zero trust security architecture",
        "autonomous robot navigation SLAM",
        "NLP sentiment analysis",
    ],
    "Natural Language": [
        "How do neural networks learn using backpropagation",
        "What are the common vulnerabilities in web applications",
        "Explain the role of smart contracts in decentralized finance",
        "How does containerization work with Docker and Kubernetes",
        "What is the difference between supervised and unsupervised learning",
    ],
    "Ambiguous": [
        "python",
        "transformer",
        "cloud",
        "security",
        "mining",
    ],
    "Noisy/Typo": [
        "machne lerning algorithams",
        "cybersecrity threts",
        "nueral netwerks nlp",
        "blockchan scalablity",
        "quntum superposishon",
    ],
}

# 
# SECTION 2 — GROUND TRUTH GENERATION
# 
# Expert-curated ground truth: maps each query to its 20 relevant document titles
MANUAL_GT_TITLES = {
    "blockchain consensus algorithms": [
        "Blockchain","Bitcoin","Bitcoin protocol","Ethereum","Cryptocurrency",
        "IOTA (technology)","Environmental impact of bitcoin",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "Blockchain-Based Federated Learning: Incentivizing Data Sharing and Penalizing Dishonest Behavior",
        "An Anonymous Trust-Marking Scheme on Blockchain Systems",
        "A trustless society? A political look at the blockchain vision",
        "Selecting Reliable Blockchain Peers via Hybrid Blockchain Reliability Prediction",
        "Advanced Drone Swarm Security by Using Blockchain Governance Game",
        "Private and Secure Post-Quantum Verifiable Random Function with NIZK Proof and Ring-LWE Encryption in Blockchain",
        "ChainSplitter: Towards Blockchain-based Industrial IoT Architecture for Supporting Hierarchical Storage",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
        "How consensus mechanisms work in blockchain",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
    ],
    "quantum entanglement computing": [
        "Quantum computing","Qubit","Quantum supremacy","Quantum network","Quantum logic gate",
        "Topological quantum computer","Glossary of quantum computing",
        "Entangling Disciplines: Causality, Entropy and Time-Travel Paradoxes on a Quantum Computer",
        "Quantum simulation and computing with Rydberg-interacting qubits",
        "Quantum Computing: Vision and Challenges","Strengths and Weaknesses of Quantum Computing",
        "Quantum Computing in the NISQ era and beyond","Quantum Algorithm Implementations for Beginners",
        "Quantum computing and information extraction for a dynamical quantum system",
        "Universality in Quantum Computation","Quantum entanglement and its role in computing",
        "Quantum Computing Might Change AI More Than GPUs Did",
        "⚛︎ Quantum Computing in Practice: Superposition, Entanglement, and Algorithms with Qiskit",
        "Quantum Computing's 2026 Reckoning: What's Real vs. Hype",
        "Quantum Computing: A Catalyst for the Future of Digital Asset Custody in Tokenization",
    ],
    "zero trust security architecture": [
        "The fundamentals of zero-trust security architecture","Computer security",
        "Outline of computer security","Information security standards",
        "Vulnerability (computer security)","Cyberattack","Security hacker",
        "Information security audit","RSA Security",
        "Enabling AutoML for Zero-Touch Network Security: Use-Case Driven Analysis",
        "Differentiated Security Architecture for Secure and Efficient Infotainment Data Communication in IoV Networks",
        "Ontologies for Network Security and Future Challenges",
        "Evaluation of Enhanced Security Solutions in 802.11-Based Networks",
        "Complete Security Framework for Wireless Sensor Networks",
        "A security suite for wireless body area networks",
        "Common vulnerabilities in web applications",
        "Post-Quantum IPsec Is Finally Becoming Boring — And That's the Point",
        "How to audit your AWS infrastructure for NIS2 and DORA compliance (practical guide)",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
    ],
    "autonomous robot navigation SLAM": [
        "How SLAM works in autonomous robots","Autonomous robot","Self-driving car",
        "Robot Operating System","Robotics","Autonomous underwater vehicle",
        "Unmanned ground vehicle","Autonomous aircraft","Vehicular automation",
        "Learning a Terrain- and Robot-Aware Dynamics Model for Autonomous Mobile Robot Navigation",
        "Planning for robotic exploration based on forward simulation",
        "On Deep Learning Techniques to Boost Monocular Depth Estimation for Autonomous Navigation",
        "Skyeye Team at MBZIRC 2020: A team of aerial and ground robots for GPS-denied autonomous fire extinguishing in an urban building scenario",
        "A Verification Methodology for Safety Assurance of Robotic Autonomous Systems",
        "The Safety Filter: A Unified View of Safety-Critical Control in Autonomous Systems",
        "Analysis and Control of Autonomous Mobility-on-Demand Systems",
        "Fault-Tolerant Multi-Robot Coordination with Limited Sensing within Confined Environments",
        "Learning agent's spatial configuration from sensorimotor invariants",
        "The role of computer vision in robotics",
        "CE-MRS: Contrastive Explanations for Multi-Robot Systems",
    ],
    "NLP sentiment analysis": [
        "How sentiment analysis works in NLP","Natural language processing","BERT (language model)",
        "Large language model","Deep learning","Outline of natural language processing","Semantic similarity",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Building a Spam Classifier using Machine Learning and Flask",
        "Your Pipeline Is 27.7h Behind: Catching Real Estate Sentiment Leads with Pulsebit",
        "Your Pipeline Is 27.4h Behind: Catching Real Estate Sentiment Leads with Pulsebit",
        "How to Detect Real Estate Sentiment Anomalies with the Pulsebit API (Python)",
        "Your Pipeline Is 24.8h Behind: Catching Artificial Intelligence Sentiment Leads with Pulsebit",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "HumanPersonaBase: A Language-Agnostic Framework for Human-like AI Communication",
        "Named entity recognition techniques",
    ],
    "How do neural networks learn using backpropagation": [
        "How neural networks learn using backpropagation","Machine learning","Outline of machine learning",
        "Deep learning","Adversarial machine learning","Support vector machine","Federated learning",
        "Feature learning","How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines","Data preprocessing and cleaning techniques",
        "BERT (language model)",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "Active learning for data streams: a survey",
        "The role of transformers in modern NLP",
        "Reinforcement learning and its real-world applications",
    ],
    "What are the common vulnerabilities in web applications": [
        "Common vulnerabilities in web applications","Vulnerability (computer security)","Computer security",
        "Cyberattack","Security hacker","Outline of computer security","Social engineering (security)",
        "How phishing attacks are carried out and prevented",
        "GHSA-46FP-8F5P-PF2M: XSS Filter Bypass via Improper HTML Entity Decoding in Loofah",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "Web Pentesting Beginner Roadmap (2026): From Recon to Server-Side Attacks",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
        "GHSA-5X2W-37XF-7962: Unauthenticated PGP Decryption and Resource Exhaustion in AVideo",
        "GHSA-4FCP-JXH7-23X8: Unbounded YAML Alias Expansion Denial of Service in Dasel",
        "I Upgraded My Pentesting Toolkit: Introducing 2 Free Web Tools for Infosec",
        "Best Cyber Range Platform for Beginners and Professionals",
        "MediaTek Audio DSP Vulnerability: How a Nothing Phone Could Have Been Hacked (Except It Wasn't)",
        "Automated Big Text Security Classification",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
    ],
    "Explain the role of smart contracts in decentralized finance": [
        "The use of smart contracts in decentralized finance","Ethereum","Blockchain",
        "Non-fungible token","Cryptocurrency",
        "SmartOil: Blockchain and smart contract-based oil supply chain management",
        "An Anonymous Trust-Marking Scheme on Blockchain Systems",
        "Advanced Drone Swarm Security by Using Blockchain Governance Game",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "How consensus mechanisms work in blockchain","Blockchain scalability challenges and solutions",
        "OWASP Smart Contract Top 10: 2026 — The Practical Defense Guide Nobody Wrote Yet",
        "The Gondi Exploit Dissected: How a Broken Authorization Check in an NFT Lending Protocol Let Anyone Walk Away With 78 NFTs",
        "AI + Foundry's cast: how I investigate smart contracts now",
        "From Web2 to Web3: A Practical Guide to Building a Gasless Crypto Payment Flow (With Starknet & StarkZap)",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
        "Ethereum Tests Native Rollups, Vitalik Pushes One-Click Staking, Native Account Abstraction, ACDE #232 Delays Hegota",
        "The Authorization Abuse Epidemic: Why Permit2 Signature Phishing Is Now Crypto's Deadliest Attack Vector",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
    ],
    "How does containerization work with Docker and Kubernetes": [
        "How containerization works with Docker and Kubernetes","Cloud computing","Amazon Web Services",
        "Microsoft Azure","Amazon Elastic Compute Cloud","OpenStack","Cloud computing security",
        "Cloud computing issues","Serverless computing explained",
        "Supporting Multi-Cloud in Serverless Computing","IAM Permissions Are Confusing",
        "Deploying with Terraform","AWS Certified DevOps Professional Training: Career Growth Guide",
        "Scaling AWS FIFO SQS (queues) without blocking customers",
        "Top 5 Multi-Cloud FinOps Challenges and How to Solve Them",
        "Master in Azure DevOps: Training, Skills, Career Path",
        "A Comparative Study of Load Balancing Algorithms in Cloud Computing Environment",
        "SLA-Oriented Resource Provisioning for Cloud Computing: Challenges, Architecture, and Solutions",
        "Application of Selective Algorithm for Effective Resource Provisioning in Cloud Computing Environment",
        "Accelerator Virtualization in Fog Computing: Moving From the Cloud to the Edge",
    ],
    "What is the difference between supervised and unsupervised learning": [
        "Machine learning","Outline of machine learning","Adversarial machine learning",
        "Support vector machine","Federated learning","Deep learning","Feature learning",
        "How neural networks learn using backpropagation","How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines","Data preprocessing and cleaning techniques",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Active learning for data streams: a survey",
        "A review of clustering models in educational data science towards fairness-aware learning",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Reinforcement learning and its real-world applications",
        "Artificial intelligence",
        "A Benchmark Study of Machine Learning Models for Online Fake News Detection",
    ],
    "python": [
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "PyThaiNLP: Thai Natural Language Processing in Python",
        "Building a Spam Classifier using Machine Learning and Flask",
        "My Recommended Resources to Learn NLP with Python",
        "From Web Table to Pandas DataFrame in 30 Seconds",
        "Cleaning Cinema Titles Before You Can Even Search",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Feature engineering in machine learning pipelines","How gradient boosting algorithms work",
        "Data preprocessing and cleaning techniques","Machine learning",
        "NER: Gemini vs Spacy vs Compromise",
        "From PDF to Markdown: Why Document Parsing is Important For RAG",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "Building CDDBS — Part 3: Scoring LLM Output Without Another LLM",
        "Natural Language Processing for Requirements Traceability",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Active learning for data streams: a survey",
        "I Built an Interactive SQL Quiz App for devs and analysts",
        "Natural language processing",
    ],
    "transformer": [
        "BERT (language model)","Large language model","The role of transformers in modern NLP",
        "Deep learning","Prompt engineering","Natural language processing",
        "Outline of artificial intelligence","Language model benchmark",
        "Building CDDBS — Part 3: Scoring LLM Output Without Another LLM",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "SpeechPrompt: Prompting Speech Language Models for Speech Processing Tasks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "Inside a Production RAG System: Architecture, Stack, and Lessons Learned",
        "How sentiment analysis works in NLP","Artificial intelligence","Artificial intelligence engineering",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Named entity recognition techniques",
    ],
    "cloud": [
        "Cloud computing","Cloud storage","Cloud computing security","Cloud computing issues",
        "Amazon Web Services","Microsoft Azure","Amazon Elastic Compute Cloud","HP Cloud",
        "OpenStack","OpenNebula",
        "A Comparative Study of Load Balancing Algorithms in Cloud Computing Environment",
        "Ad hoc Cloud Computing: From Concept to Realization",
        "Supporting Multi-Cloud in Serverless Computing",
        "Framework for cloud computing adoption: A road map for Smes to cloud migration",
        "SLA-Oriented Resource Provisioning for Cloud Computing: Challenges, Architecture, and Solutions",
        "Application of Selective Algorithm for Effective Resource Provisioning in Cloud Computing Environment",
        "Emerging Security Challenges of Cloud Virtual Infrastructure",
        "How containerization works with Docker and Kubernetes",
        "Serverless computing explained","Cloud storage architecture and reliability",
    ],
    "security": [
        "Computer security","Information security standards","Cyberattack",
        "Vulnerability (computer security)","Security hacker","Outline of computer security",
        "Social engineering (security)","Information security audit","RSA Security",
        "How phishing attacks are carried out and prevented",
        "The fundamentals of zero-trust security architecture",
        "Common vulnerabilities in web applications",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
        "Enabling AutoML for Zero-Touch Network Security: Use-Case Driven Analysis",
        "Ontologies for Network Security and Future Challenges",
        "Complete Security Framework for Wireless Sensor Networks",
        "A security suite for wireless body area networks",
    ],
    "mining": [
        "Bitcoin","Bitcoin protocol","Environmental impact of bitcoin","Blockchain","Cryptocurrency",
        "Cryptocurrency and crime","Cryptocurrency bubble","IOTA (technology)",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Bitcoin's Edge: Embedded Sentiment in Blockchain Transactional Data",
        "How consensus mechanisms work in blockchain","Blockchain scalability challenges and solutions",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
        "GlassWorm's Solana C2: How a Supply-Chain Monster Turned the Blockchain Into a Dead Drop",
        "Solana Memo Program as Malware C2: Inside the GlassWorm Supply Chain Attack That Weaponized the Blockchain",
        "Address Poisoning After Fusaka: How Ethereum's Fee Cut Handed Scammers a 612% Boost",
        "The Authorization Abuse Epidemic: Why Permit2 Signature Phishing Is Now Crypto's Deadliest Attack Vector",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
    ],
    "machne lerning algorithams": [
        "Machine learning","Outline of machine learning","Adversarial machine learning",
        "Support vector machine","Deep learning","Federated learning","Feature learning",
        "How neural networks learn using backpropagation","How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Active learning for data streams: a survey",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "ALERT-Transformer: Bridging Asynchronous and Synchronous Machine Learning for Real-Time Event-based Spatio-Temporal Data",
        "A Perspective on Symbolic Machine Learning in Physical Sciences",
        "The Scientific Method in the Science of Machine Learning",
        "Artificial intelligence",
        "A Benchmark Study of Machine Learning Models for Online Fake News Detection",
    ],
    "cybersecrity threts": [
        "Cyberattack","Social engineering (security)","Vulnerability (computer security)",
        "Computer security","Security hacker",
        "How phishing attacks are carried out and prevented","Common vulnerabilities in web applications",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "MediaTek Audio DSP Vulnerability: How a Nothing Phone Could Have Been Hacked (Except It Wasn't)",
        "GHSA-5X2W-37XF-7962: Unauthenticated PGP Decryption and Resource Exhaustion in AVideo",
        "GHSA-4FCP-JXH7-23X8: Unbounded YAML Alias Expansion Denial of Service in Dasel",
        "GHSA-46FP-8F5P-PF2M: XSS Filter Bypass via Improper HTML Entity Decoding in Loofah",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "Web Pentesting Beginner Roadmap (2026): From Recon to Server-Side Attacks",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
        "Automated Big Text Security Classification",
        "Securing Automated Insulin Delivery Systems: A Review of Security Threats and Protective Strategies",
        "Developments in Connected Vehicles and the Requirement for Increased Cybersecurity",
        "Ontologies for Network Security and Future Challenges",
        "Autonomous Low Power IoT System Architecture for Cybersecurity Monitoring",
        "The fundamentals of zero-trust security architecture",
        "Enabling AutoML for Zero-Touch Network Security: Use-Case Driven Analysis",
        "Complete Security Framework for Wireless Sensor Networks",
        "A security suite for wireless body area networks",
    ],
    "nueral netwerks nlp": [
        "BERT (language model)","Large language model","Deep learning","Natural language processing",
        "Outline of artificial intelligence","Artificial intelligence",
        "The role of transformers in modern NLP","How neural networks learn using backpropagation",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "How sentiment analysis works in NLP","Named entity recognition techniques",
        "Machine translation and its challenges",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "SpeechPrompt: Prompting Speech Language Models for Speech Processing Tasks",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Building a Spam Classifier using Machine Learning and Flask",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "From Statistical Relational to Neurosymbolic Artificial Intelligence: a Survey",
    ],
    "blockchan scalablity": [
        "Blockchain scalability challenges and solutions","Blockchain","Ethereum","Bitcoin",
        "IOTA (technology)",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "ChainSplitter: Towards Blockchain-based Industrial IoT Architecture for Supporting Hierarchical Storage",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
        "Blockchain-Based Federated Learning: Incentivizing Data Sharing and Penalizing Dishonest Behavior",
        "Selecting Reliable Blockchain Peers via Hybrid Blockchain Reliability Prediction",
        "How consensus mechanisms work in blockchain","The use of smart contracts in decentralized finance",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
        "Ethereum Tests Native Rollups, Vitalik Pushes One-Click Staking, Native Account Abstraction, ACDE #232 Delays Hegota",
        "From Web2 to Web3: A Practical Guide to Building a Gasless Crypto Payment Flow (With Starknet & StarkZap)",
        "Digital Twins and Blockchain for IoT Management",
        "Customer Data Sharing Platform: A Blockchain-Based Shopping Cart",
    ],
    "quntum superposishon": [
        "Quantum computing","Qubit","Quantum logic gate","Topological quantum computer",
        "Linear optical quantum computing","One-way quantum computer","Quantum supremacy",
        "Glossary of quantum computing","Quantum Algorithm Implementations for Beginners",
        "Quantum Computing: Vision and Challenges","Strengths and Weaknesses of Quantum Computing",
        "Quantum Computing in the NISQ era and beyond","Universality in Quantum Computation",
        "Dissipation as a resource for Quantum Reservoir Computing",
        "Quantum computing and information extraction for a dynamical quantum system",
        "Quantum entanglement and its role in computing","How quantum error correction works",
        "⚛︎ Quantum Computing in Practice: Superposition, Entanglement, and Algorithms with Qiskit",
        "Quantum Computing Might Change AI More Than GPUs Did",
        "Quantum Computing's 2026 Reckoning: What's Real vs. Hype",
    ],
}

def build_ground_truth(system, queries_dict):
    """Build ground truth from expert-curated document titles for all 20 queries."""
    title_to_idx = {t: i for i, t in enumerate(system.df["title"])}
    ground_truth = {}
    for category, queries in queries_dict.items():
        for query in queries:
            titles = MANUAL_GT_TITLES.get(query, [])
            ground_truth[query] = [title_to_idx[t] for t in titles if t in title_to_idx]
    return ground_truth

# 
# SECTION 3 — METRICS
# 
def precision_at_k(retrieved_indices, relevant_indices, k=10):
    if not retrieved_indices: return 0.0
    retrieved_k = retrieved_indices[:k]
    relevant_set = set(relevant_indices)
    intersection = [idx for idx in retrieved_k if idx in relevant_set]
    return len(intersection) / k

def recall(retrieved_indices, relevant_indices):
    if not relevant_indices: return 0.0
    relevant_set = set(relevant_indices)
    intersection = [idx for idx in retrieved_indices if idx in relevant_set]
    return len(intersection) / len(relevant_indices)

def average_precision(retrieved_indices, relevant_indices):
    relevant_set = set(relevant_indices)
    ap = 0.0
    relevant_count = 0
    for i, idx in enumerate(retrieved_indices):
        if idx in relevant_set:
            relevant_count += 1
            ap += relevant_count / (i + 1)
    if not relevant_indices: return 0.0
    return ap / len(relevant_indices)

# 
# SECTION 4 — EVALUATION RUNNER
# 
def run_evaluation():
    print("Loading retrieval system for evaluation...")
    system = RetrievalSystem("corpus.csv")
    ground_truth = build_ground_truth(system, TEST_QUERIES)
    
    models = ["VSM", "BM25", "Embedding"]
    results = []

    print("\nRunning evaluation on 20 queries...")
    for category, queries in TEST_QUERIES.items():
        for query in queries:
            processed_data = preprocess_query(query)
            query_str = processed_data['processed']
            is_phrase = processed_data['is_phrase']
            rel_indices = ground_truth[query]
            
            for model in models:
                search_results = system.search(query_str, model_name=model, is_phrase=is_phrase, top_k=50)
                retrieved_indices = [r['doc_index'] for r in search_results]
                
                p10 = precision_at_k(retrieved_indices, rel_indices, k=10)
                rec = recall(retrieved_indices, rel_indices)
                ap = average_precision(retrieved_indices, rel_indices)
                
                results.append({
                    "Category": category,
                    "Query": query,
                    "Model": model,
                    "P@10": p10,
                    "Recall": rec,
                    "AP": ap
                })
    
    df_results = pd.DataFrame(results)
    
    summary = df_results.groupby(["Category", "Model"])[["P@10", "Recall", "AP"]].mean().reset_index()
    print("\n--- Evaluation Summary (Averaged by Category) ---")
    print(summary.to_string(index=False))
    
    overall = df_results.groupby("Model")[["P@10", "Recall", "AP"]].mean().reset_index()
    overall.rename(columns={"AP": "MAP"}, inplace=True)
    print("\n--- Overall Model Performance (MAP) ---")
    print(overall.to_string(index=False))
    
    return df_results, summary

if __name__ == "__main__":
    run_evaluation()
```

## Query_processor

```python
import re
from spellchecker import SpellChecker


ABBREVIATIONS = pd.read_csv("abbreviations_list.csv")
ABBREVIATIONS = dict(zip(ABBREVIATIONS['Abbreviation'], ABBREVIATIONS['Expansion']))

_spell = SpellChecker()
with open('technical_terms.txt', 'r', encoding='utf-8') as f:
    technical_terms = [line.strip() for line in f if line.strip()]
    
_spell.word_frequency.load_words(technical_terms)



# Public API
def preprocess_query(query: str) -> dict:
    original = query.strip()

    # 1. Phrase detection — check for surrounding quotes BEFORE lowering
    is_phrase = False
    working = original
    if (working.startswith('"') and working.endswith('"')) or \
       (working.startswith("'") and working.endswith("'")):
        is_phrase = True
        working = working[1:-1].strip()  # strip the quotes

    # 2. Abbreviation expansion (case-sensitive match on the original tokens)
    tokens = working.split()
    expanded_tokens = []
    for token in tokens:
        # Try exact match first, then uppercase match
        clean = token.strip(".,!?;:()")
        if clean in ABBREVIATIONS:
            expanded_tokens.append(ABBREVIATIONS[clean])
        elif clean.upper() in ABBREVIATIONS:
            expanded_tokens.append(ABBREVIATIONS[clean.upper()])
        else:
            expanded_tokens.append(token)
    working = " ".join(expanded_tokens)

    # 3. Case normalization
    working = working.lower()

    # 4. Spelling correction
    words = working.split()
    corrections = {}
    corrected_words = []
    for word in words:
        # Skip short words, numbers, and already-known words
        clean_word = re.sub(r'[^\w]', '', word)
        if len(clean_word) <= 2 or clean_word.isdigit():
            corrected_words.append(word)
            continue

        if clean_word in _spell:
            corrected_words.append(word)
        else:
            suggestion = _spell.correction(clean_word)
            if suggestion and suggestion != clean_word:
                corrections[clean_word] = suggestion
                # Replace the misspelled portion, keep punctuation
                corrected_words.append(word.replace(clean_word, suggestion))
            else:
                corrected_words.append(word)

    processed = " ".join(corrected_words)

    return {
        "original": original,
        "processed": processed,
        "is_phrase": is_phrase,
        "corrections": corrections,
    }


# test
if __name__ == "__main__":
    test_cases = [
        'AI applications in healthcare',
        '"natural language processing"',
        'machne lerning algorithms',
        'NLP and SLAM techniques',
        'BERT vs GPT comparison',
        'blockchan smart contracs',
    ]
    for q in test_cases:
        result = preprocess_query(q)
        print(f"\n  Input:       {result['original']}")
        print(f"  Processed:   {result['processed']}")
        print(f"  Is phrase:   {result['is_phrase']}")
        if result['corrections']:
            print(f"  Corrections: {result['corrections']}")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer


def _tokenize(text: str) -> list:
    return re.findall(r'\w+', text.lower())


# Ranks documents using TF-IDF weighted vectors and cosine similarity
class VSMRetriever:
    def __init__(self, documents: list):
        self.documents = documents
        self.vectorizer = TfidfVectorizer(
            stop_words='english', max_features=50000, sublinear_tf=True,
        )
        self.tfidf_matrix = self.vectorizer.fit_transform(documents)
        print(f"  [VSM] TF-IDF matrix shape: {self.tfidf_matrix.shape}")

    def retrieve(self, query: str, is_phrase: bool = False, top_k: int = 10):
        # Return top-k (doc_index, score) pairs sorted descending
        query_vec = self.vectorizer.transform([query])
        scores = cosine_similarity(query_vec, self.tfidf_matrix).flatten()
        if is_phrase:
            phrase_lower = query.lower()
            for i, doc in enumerate(self.documents):
                if phrase_lower in doc.lower():
                    scores[i] *= 2.0
        top_indices = np.argsort(scores)[::-1][:top_k]
        return [(int(idx), float(scores[idx])) for idx in top_indices]


# Ranks documents using the Okapi BM25 algorithm
class BM25Retriever:
    def __init__(self, documents: list):
        self.documents = documents
        self.tokenized_docs = [_tokenize(doc) for doc in documents]
        self.bm25 = BM25Okapi(self.tokenized_docs)
        print(f"  [BM25] Indexed {len(self.tokenized_docs)} documents")

    def retrieve(self, query: str, is_phrase: bool = False, top_k: int = 10):
        # Return top-k (doc_index, score) pairs sorted descending.
        query_tokens = _tokenize(query)
        scores = self.bm25.get_scores(query_tokens)
        if is_phrase:
            phrase_lower = query.lower()
            for i, doc in enumerate(self.documents):
                if phrase_lower in doc.lower():
                    scores[i] *= 2.0
        top_indices = np.argsort(scores)[::-1][:top_k]
        return [(int(idx), float(scores[idx])) for idx in top_indices]


# Ranks documents by cosine similarity of sentence-transformer embeddings
class EmbeddingRetriever:
    def __init__(self, documents: list):
        self.documents = documents
        print("  [Embedding] Loading sentence-transformer model...")
        self.model = SentenceTransformer('all-MiniLM-L6-v2')
        truncated = [doc[:2000] for doc in documents]
        print(f"  [Embedding] Encoding {len(truncated)} documents...")
        self.doc_embeddings = self.model.encode(
            truncated, show_progress_bar=True, batch_size=64, convert_to_numpy=True,
        )
        norms = np.linalg.norm(self.doc_embeddings, axis=1, keepdims=True)
        norms[norms == 0] = 1
        self.doc_embeddings = self.doc_embeddings / norms
        print(f"  [Embedding] Embeddings shape: {self.doc_embeddings.shape}")

    def retrieve(self, query: str, is_phrase: bool = False, top_k: int = 10):
        # Return top-k (doc_index, score) pairs sorted descending
        query_emb = self.model.encode([query], convert_to_numpy=True)
        query_emb = query_emb / np.linalg.norm(query_emb)
        scores = np.dot(self.doc_embeddings, query_emb.T).flatten()
        if is_phrase:
            phrase_lower = query.lower()
            for i, doc in enumerate(self.documents):
                if phrase_lower in doc.lower():
                    scores[i] *= 1.5
        top_indices = np.argsort(scores)[::-1][:top_k]
        return [(int(idx), float(scores[idx])) for idx in top_indices]


# Wraps all three retrieval models and the corpus DataFrame
class RetrievalSystem:
    def __init__(self, corpus_path: str = "corpus.csv"):
        print("=" * 60)
        print("Initializing Retrieval System")
        print("=" * 60)
        self.df = pd.read_csv(corpus_path)
        self.df['text'] = self.df['text'].fillna('')
        self.df['title'] = self.df['title'].fillna('')
        self.documents = (self.df['title'] + " " + self.df['text']).tolist()
        print(f"\nCorpus loaded: {len(self.documents)} documents")
        print(f"Topics: {self.df['topic'].unique().tolist()}\n")

        print("Initializing VSM (TF-IDF)...")
        self.vsm = VSMRetriever(self.documents)
        print("\nInitializing BM25...")
        self.bm25 = BM25Retriever(self.documents)
        print("\nInitializing Embedding model...")
        self.embedding = EmbeddingRetriever(self.documents)
        print("\n" + "=" * 60)
        print("All models ready!")
        print("=" * 60)

    def search(self, query, model_name="VSM", is_phrase=False, top_k=10):
        # Search the corpus using the specified model. Returns list of result dicts
        model_map = {"VSM": self.vsm, "BM25": self.bm25, "Embedding": self.embedding}
        retriever = model_map.get(model_name, self.vsm)
        raw_results = retriever.retrieve(query, is_phrase=is_phrase, top_k=top_k)

        results = []
        for rank, (doc_idx, score) in enumerate(raw_results, start=1):
            row = self.df.iloc[doc_idx]
            text = str(row['text'])
            snippet = text[:200].strip()
            if len(text) > 200:
                snippet += "..."
            results.append({
                "rank": rank, "doc_index": doc_idx,
                "title": str(row['title']), "url": str(row['url']),
                "snippet": snippet, "score": round(score, 6),
                "topic": str(row['topic']), "content_type": str(row['content_type']),
            })
        return results


if __name__ == "__main__":
    system = RetrievalSystem("corpus.csv")
    test_query = "machine learning algorithms"
    for model_name in ["VSM", "BM25", "Embedding"]:
        print(f"\n{'='*60}")
        print(f"Model: {model_name} | Query: '{test_query}'")
        print(f"{'='*60}")
        results = system.search(test_query, model_name=model_name, top_k=5)
        for r in results:
            print(f"  #{r['rank']} [{r['score']:.4f}] {r['title'][:60]}  ({r['topic']})")
```

```python
#  Metric functions 
def p_at_k(retrieved, relevant, k=10):
    rs = set(relevant)
    return sum(1 for i in retrieved[:k] if i in rs) / k

def recall_sc(retrieved, relevant):
    if not relevant: return 0.0
    rs = set(relevant)
    return sum(1 for i in retrieved if i in rs) / len(relevant)

def avg_prec(retrieved, relevant):
    if not relevant: return 0.0
    rs = set(relevant); hits = ap = 0.0
    for rank, idx in enumerate(retrieved, 1):
        if idx in rs:
            hits += 1; ap += hits / rank
    return ap / len(relevant)

print("Metrics defined.")
```

```python
# Manual ground truth: expert-curated relevant document titles per query
MANUAL_GT = {
    "blockchain consensus algorithms": [
        "Blockchain", "Bitcoin", "Bitcoin protocol", "Ethereum", "Cryptocurrency",
        "IOTA (technology)", "Environmental impact of bitcoin",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "Blockchain-Based Federated Learning: Incentivizing Data Sharing and Penalizing Dishonest Behavior",
        "An Anonymous Trust-Marking Scheme on Blockchain Systems",
        "A trustless society? A political look at the blockchain vision",
        "Selecting Reliable Blockchain Peers via Hybrid Blockchain Reliability Prediction",
        "Advanced Drone Swarm Security by Using Blockchain Governance Game",
        "Private and Secure Post-Quantum Verifiable Random Function with NIZK Proof and Ring-LWE Encryption in Blockchain",
        "ChainSplitter: Towards Blockchain-based Industrial IoT Architecture for Supporting Hierarchical Storage",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
        "How consensus mechanisms work in blockchain",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
    ],
    "quantum entanglement computing": [
        "Quantum computing", "Qubit", "Quantum supremacy", "Quantum network",
        "Quantum logic gate", "Topological quantum computer", "Glossary of quantum computing",
        "Entangling Disciplines: Causality, Entropy and Time-Travel Paradoxes on a Quantum Computer",
        "Quantum simulation and computing with Rydberg-interacting qubits",
        "Quantum Computing: Vision and Challenges", "Strengths and Weaknesses of Quantum Computing",
        "Quantum Computing in the NISQ era and beyond",
        "Quantum Algorithm Implementations for Beginners",
        "Quantum computing and information extraction for a dynamical quantum system",
        "Universality in Quantum Computation",
        "Quantum entanglement and its role in computing",
        "Quantum Computing Might Change AI More Than GPUs Did",
        "⛛︎ Quantum Computing in Practice: Superposition, Entanglement, and Algorithms with Qiskit",
        "Quantum Computing's 2026 Reckoning: What's Real vs. Hype",
        "Quantum Computing: A Catalyst for the Future of Digital Asset Custody in Tokenization",
    ],
    "zero trust security architecture": [
        "The fundamentals of zero-trust security architecture", "Computer security",
        "Outline of computer security", "Information security standards",
        "Vulnerability (computer security)", "Cyberattack", "Security hacker",
        "Information security audit", "RSA Security",
        "Enabling AutoML for Zero-Touch Network Security: Use-Case Driven Analysis",
        "Differentiated Security Architecture for Secure and Efficient Infotainment Data Communication in IoV Networks",
        "Ontologies for Network Security and Future Challenges",
        "Evaluation of Enhanced Security Solutions in 802.11-Based Networks",
        "Complete Security Framework for Wireless Sensor Networks",
        "A security suite for wireless body area networks",
        "Common vulnerabilities in web applications",
        "Post-Quantum IPsec Is Finally Becoming Boring — And That's the Point",
        "How to audit your AWS infrastructure for NIS2 and DORA compliance (practical guide)",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
    ],
    "autonomous robot navigation SLAM": [
        "How SLAM works in autonomous robots", "Autonomous robot", "Self-driving car",
        "Robot Operating System", "Robotics", "Autonomous underwater vehicle",
        "Unmanned ground vehicle", "Autonomous aircraft", "Vehicular automation",
        "Learning a Terrain- and Robot-Aware Dynamics Model for Autonomous Mobile Robot Navigation",
        "Planning for robotic exploration based on forward simulation",
        "On Deep Learning Techniques to Boost Monocular Depth Estimation for Autonomous Navigation",
        "Skyeye Team at MBZIRC 2020: A team of aerial and ground robots for GPS-denied autonomous fire extinguishing in an urban building scenario",
        "A Verification Methodology for Safety Assurance of Robotic Autonomous Systems",
        "The Safety Filter: A Unified View of Safety-Critical Control in Autonomous Systems",
        "Analysis and Control of Autonomous Mobility-on-Demand Systems",
        "Fault-Tolerant Multi-Robot Coordination with Limited Sensing within Confined Environments",
        "Learning agent's spatial configuration from sensorimotor invariants",
        "The role of computer vision in robotics",
        "CE-MRS: Contrastive Explanations for Multi-Robot Systems",
    ],
    "NLP sentiment analysis": [
        "How sentiment analysis works in NLP", "Natural language processing",
        "BERT (language model)", "Large language model", "Deep learning",
        "Outline of natural language processing", "Semantic similarity",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Building a Spam Classifier using Machine Learning and Flask",
        "Your Pipeline Is 27.7h Behind: Catching Real Estate Sentiment Leads with Pulsebit",
        "Your Pipeline Is 27.4h Behind: Catching Real Estate Sentiment Leads with Pulsebit",
        "How to Detect Real Estate Sentiment Anomalies with the Pulsebit API (Python)",
        "Your Pipeline Is 24.8h Behind: Catching Artificial Intelligence Sentiment Leads with Pulsebit",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "HumanPersonaBase: A Language-Agnostic Framework for Human-like AI Communication",
        "Named entity recognition techniques",
    ],
    "How do neural networks learn using backpropagation": [
        "How neural networks learn using backpropagation", "Machine learning",
        "Outline of machine learning", "Deep learning", "Adversarial machine learning",
        "Support vector machine", "Federated learning", "Feature learning",
        "How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines",
        "Data preprocessing and cleaning techniques", "BERT (language model)",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "Active learning for data streams: a survey",
        "The role of transformers in modern NLP",
        "Reinforcement learning and its real-world applications",
    ],
    "What are the common vulnerabilities in web applications": [
        "Common vulnerabilities in web applications", "Vulnerability (computer security)",
        "Computer security", "Cyberattack", "Security hacker",
        "Outline of computer security", "Social engineering (security)",
        "How phishing attacks are carried out and prevented",
        "GHSA-46FP-8F5P-PF2M: XSS Filter Bypass via Improper HTML Entity Decoding in Loofah",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "Web Pentesting Beginner Roadmap (2026): From Recon to Server-Side Attacks",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
        "GHSA-5X2W-37XF-7962: Unauthenticated PGP Decryption and Resource Exhaustion in AVideo",
        "GHSA-4FCP-JXH7-23X8: Unbounded YAML Alias Expansion Denial of Service in Dasel",
        "I Upgraded My Pentesting Toolkit: Introducing 2 Free Web Tools for Infosec",
        "Best Cyber Range Platform for Beginners and Professionals",
        "MediaTek Audio DSP Vulnerability: How a Nothing Phone Could Have Been Hacked (Except It Wasn't)",
        "Automated Big Text Security Classification",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
    ],
    "Explain the role of smart contracts in decentralized finance": [
        "The use of smart contracts in decentralized finance", "Ethereum", "Blockchain",
        "Non-fungible token", "Cryptocurrency",
        "SmartOil: Blockchain and smart contract-based oil supply chain management",
        "An Anonymous Trust-Marking Scheme on Blockchain Systems",
        "Advanced Drone Swarm Security by Using Blockchain Governance Game",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "How consensus mechanisms work in blockchain",
        "Blockchain scalability challenges and solutions",
        "OWASP Smart Contract Top 10: 2026 — The Practical Defense Guide Nobody Wrote Yet",
        "The Gondi Exploit Dissected: How a Broken Authorization Check in an NFT Lending Protocol Let Anyone Walk Away With 78 NFTs",
        "AI + Foundry's cast: how I investigate smart contracts now",
        "From Web2 to Web3: A Practical Guide to Building a Gasless Crypto Payment Flow (With Starknet & StarkZap)",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
        "Ethereum Tests Native Rollups, Vitalik Pushes One-Click Staking, Native Account Abstraction, ACDE #232 Delays Hegota",
        "The Authorization Abuse Epidemic: Why Permit2 Signature Phishing Is Now Crypto's Deadliest Attack Vector",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
    ],
    "How does containerization work with Docker and Kubernetes": [
        "How containerization works with Docker and Kubernetes", "Cloud computing",
        "Amazon Web Services", "Microsoft Azure", "Amazon Elastic Compute Cloud",
        "OpenStack", "Cloud computing security", "Cloud computing issues",
        "Serverless computing explained", "Supporting Multi-Cloud in Serverless Computing",
        "IAM Permissions Are Confusing", "Deploying with Terraform",
        "AWS Certified DevOps Professional Training: Career Growth Guide",
        "Scaling AWS FIFO SQS (queues) without blocking customers",
        "Top 5 Multi-Cloud FinOps Challenges and How to Solve Them",
        "Master in Azure DevOps: Training, Skills, Career Path",
        "A Comparative Study of Load Balancing Algorithms in Cloud Computing Environment",
        "SLA-Oriented Resource Provisioning for Cloud Computing: Challenges, Architecture, and Solutions",
        "Application of Selective Algorithm for Effective Resource Provisioning in Cloud Computing Environment",
        "Accelerator Virtualization in Fog Computing: Moving From the Cloud to the Edge",
    ],
    "What is the difference between supervised and unsupervised learning": [
        "Machine learning", "Outline of machine learning", "Adversarial machine learning",
        "Support vector machine", "Federated learning", "Deep learning", "Feature learning",
        "How neural networks learn using backpropagation",
        "How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines",
        "Data preprocessing and cleaning techniques",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Active learning for data streams: a survey",
        "A review of clustering models in educational data science towards fairness-aware learning",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Reinforcement learning and its real-world applications",
        "Artificial intelligence",
        "A Benchmark Study of Machine Learning Models for Online Fake News Detection",
    ],
    "python": [
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "PyThaiNLP: Thai Natural Language Processing in Python",
        "Building a Spam Classifier using Machine Learning and Flask",
        "My Recommended Resources to Learn NLP with Python",
        "From Web Table to Pandas DataFrame in 30 Seconds",
        "Cleaning Cinema Titles Before You Can Even Search",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Feature engineering in machine learning pipelines",
        "How gradient boosting algorithms work",
        "Data preprocessing and cleaning techniques",
        "Machine learning",
        "NER: Gemini vs Spacy vs Compromise",
        "From PDF to Markdown: Why Document Parsing is Important For RAG",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "Building CDDBS — Part 3: Scoring LLM Output Without Another LLM",
        "Natural Language Processing for Requirements Traceability",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Active learning for data streams: a survey",
        "I Built an Interactive SQL Quiz App for devs and analysts",
        "Natural language processing",
    ],
    "transformer": [
        "BERT (language model)", "Large language model",
        "The role of transformers in modern NLP", "Deep learning", "Prompt engineering",
        "Natural language processing", "Outline of artificial intelligence",
        "Language model benchmark",
        "Building CDDBS — Part 3: Scoring LLM Output Without Another LLM",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "SpeechPrompt: Prompting Speech Language Models for Speech Processing Tasks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "Inside a Production RAG System: Architecture, Stack, and Lessons Learned",
        "How sentiment analysis works in NLP", "Artificial intelligence",
        "Artificial intelligence engineering",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Named entity recognition techniques",
    ],
    "cloud": [
        "Cloud computing", "Cloud storage", "Cloud computing security",
        "Cloud computing issues", "Amazon Web Services", "Microsoft Azure",
        "Amazon Elastic Compute Cloud", "HP Cloud", "OpenStack", "OpenNebula",
        "A Comparative Study of Load Balancing Algorithms in Cloud Computing Environment",
        "Ad hoc Cloud Computing: From Concept to Realization",
        "Supporting Multi-Cloud in Serverless Computing",
        "Framework for cloud computing adoption: A road map for Smes to cloud migration",
        "SLA-Oriented Resource Provisioning for Cloud Computing: Challenges, Architecture, and Solutions",
        "Application of Selective Algorithm for Effective Resource Provisioning in Cloud Computing Environment",
        "Emerging Security Challenges of Cloud Virtual Infrastructure",
        "How containerization works with Docker and Kubernetes",
        "Serverless computing explained",
        "Cloud storage architecture and reliability",
    ],
    "security": [
        "Computer security", "Information security standards", "Cyberattack",
        "Vulnerability (computer security)", "Security hacker",
        "Outline of computer security", "Social engineering (security)",
        "Information security audit", "RSA Security",
        "How phishing attacks are carried out and prevented",
        "The fundamentals of zero-trust security architecture",
        "Common vulnerabilities in web applications",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "What is ModSecurity and How It Works",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
        "Enabling AutoML for Zero-Touch Network Security: Use-Case Driven Analysis",
        "Ontologies for Network Security and Future Challenges",
        "Complete Security Framework for Wireless Sensor Networks",
        "A security suite for wireless body area networks",
    ],
    "mining": [
        "Bitcoin", "Bitcoin protocol", "Environmental impact of bitcoin",
        "Blockchain", "Cryptocurrency", "Cryptocurrency and crime",
        "Cryptocurrency bubble", "IOTA (technology)",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Bitcoin's Edge: Embedded Sentiment in Blockchain Transactional Data",
        "How consensus mechanisms work in blockchain",
        "Blockchain scalability challenges and solutions",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
        "GlassWorm's Solana C2: How a Supply-Chain Monster Turned the Blockchain Into a Dead Drop",
        "Solana Memo Program as Malware C2: Inside the GlassWorm Supply Chain Attack That Weaponized the Blockchain",
        "Address Poisoning After Fusaka: How Ethereum's Fee Cut Handed Scammers a 612% Boost",
        "The Authorization Abuse Epidemic: Why Permit2 Signature Phishing Is Now Crypto's Deadliest Attack Vector",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
    ],
    "machne lerning algorithams": [
        "Machine learning", "Outline of machine learning", "Adversarial machine learning",
        "Support vector machine", "Deep learning", "Federated learning", "Feature learning",
        "How neural networks learn using backpropagation",
        "How gradient boosting algorithms work",
        "Feature engineering in machine learning pipelines",
        "Machine Learning in Python: Main developments and technology trends in data science, machine learning, and artificial intelligence",
        "Learning Curves for Decision Making in Supervised Machine Learning: A Survey",
        "Physics-Inspired Interpretability Of Machine Learning Models",
        "Active learning for data streams: a survey",
        "DOME: Recommendations for supervised machine learning validation in biology",
        "ALERT-Transformer: Bridging Asynchronous and Synchronous Machine Learning for Real-Time Event-based Spatio-Temporal Data",
        "A Perspective on Symbolic Machine Learning in Physical Sciences",
        "The Scientific Method in the Science of Machine Learning",
        "Artificial intelligence",
        "A Benchmark Study of Machine Learning Models for Online Fake News Detection",
    ],
    "cybersecrity threts": [
        "Cyberattack", "Social engineering (security)", "Vulnerability (computer security)",
        "Computer security", "Security hacker",
        "How phishing attacks are carried out and prevented",
        "Common vulnerabilities in web applications",
        "CVE-2026-33017: How I Found an Unauthenticated RCE in Langflow by Reading the Code They Already Fixed",
        "MediaTek Audio DSP Vulnerability: How a Nothing Phone Could Have Been Hacked (Except It Wasn't)",
        "GHSA-5X2W-37XF-7962: Unauthenticated PGP Decryption and Resource Exhaustion in AVideo",
        "GHSA-4FCP-JXH7-23X8: Unbounded YAML Alias Expansion Denial of Service in Dasel",
        "GHSA-46FP-8F5P-PF2M: XSS Filter Bypass via Improper HTML Entity Decoding in Loofah",
        "What Is Remote Code Execution (RCE)? How It Works, Risks & Prevention Guide",
        "Web Pentesting Beginner Roadmap (2026): From Recon to Server-Side Attacks",
        "Classifying SMEs for Approaching Cybersecurity Competence and Awareness",
        "Automated Big Text Security Classification",
        "Securing Automated Insulin Delivery Systems: A Review of Security Threats and Protective Strategies",
        "Developments in Connected Vehicles and the Requirement for Increased Cybersecurity",
        "Ontologies for Network Security and Future Challenges",
        "Autonomous Low Power IoT System Architecture for Cybersecurity Monitoring",
    ],
    "nueral netwerks nlp": [
        "BERT (language model)", "Large language model", "Deep learning",
        "Natural language processing", "Outline of artificial intelligence",
        "Artificial intelligence", "The role of transformers in modern NLP",
        "How neural networks learn using backpropagation",
        "This Reads Like That: Deep Learning for Interpretable Natural Language Processing",
        "Image-based Natural Language Understanding Using 2D Convolutional Neural Networks",
        "Modular Mechanistic Networks: On Bridging Mechanistic and Phenomenological Models with Deep Neural Networks in Natural Language Processing",
        "How sentiment analysis works in NLP", "Named entity recognition techniques",
        "Machine translation and its challenges",
        "Comparative Analysis of Word Embeddings for Capturing Word Similarities",
        "SpeechPrompt: Prompting Speech Language Models for Speech Processing Tasks",
        "Building an AI-Generated Text Detector: A Full-Stack NLP Project Guide",
        "Building a Spam Classifier using Machine Learning and Flask",
        "Context-Anchored Generation (CAG): Fixing Hallucinations at the Decoding Layer",
        "From Statistical Relational to Neurosymbolic Artificial Intelligence: a Survey",
    ],
    "blockchan scalablity": [
        "Blockchain scalability challenges and solutions", "Blockchain", "Ethereum",
        "Bitcoin", "IOTA (technology)",
        "BlockSim: An Extensible Simulation Tool for Blockchain Systems",
        "Oceanic Games: Centralization Risks and Incentives in Blockchain Mining",
        "ChainSplitter: Towards Blockchain-based Industrial IoT Architecture for Supporting Hierarchical Storage",
        "Blockchain for Business Process Enactment: A Taxonomy and Systematic Literature Review",
        "Blockchain-Based Federated Learning: Incentivizing Data Sharing and Penalizing Dishonest Behavior",
        "Selecting Reliable Blockchain Peers via Hybrid Blockchain Reliability Prediction",
        "How consensus mechanisms work in blockchain",
        "The use of smart contracts in decentralized finance",
        "Solana's Alpenglow Security Trade-Off: How Dropping PoH for 150ms Finality Changes Every Assumption DeFi Developers Hold",
        "Firedancer's Double-Edged Sword: How Solana's Multi-Client Future Creates a New Class of Consensus-Splitting Vulnerabilities",
        "The Solana CPI Security Playbook: 7 Cross-Program Invocation Patterns That Prevent Nine-Figure Exploits",
        "Ethereum Tests Native Rollups, Vitalik Pushes One-Click Staking, Native Account Abstraction, ACDE #232 Delays Hegota",
        "From Web2 to Web3: A Practical Guide to Building a Gasless Crypto Payment Flow (With Starknet & StarkZap)",
        "Digital Twins and Blockchain for IoT Management",
        "Customer Data Sharing Platform: A Blockchain-Based Shopping Cart",
    ],
    "quntum superposishon": [
        "Quantum computing", "Qubit", "Quantum logic gate", "Topological quantum computer",
        "Linear optical quantum computing", "One-way quantum computer", "Quantum supremacy",
        "Glossary of quantum computing",
        "Quantum Algorithm Implementations for Beginners",
        "Quantum Computing: Vision and Challenges",
        "Strengths and Weaknesses of Quantum Computing",
        "Quantum Computing in the NISQ era and beyond",
        "Universality in Quantum Computation",
        "Dissipation as a resource for Quantum Reservoir Computing",
        "Quantum computing and information extraction for a dynamical quantum system",
        "Quantum entanglement and its role in computing",
        "How quantum error correction works",
        "⛛︎ Quantum Computing in Practice: Superposition, Entanglement, and Algorithms with Qiskit",
        "Quantum Computing Might Change AI More Than GPUs Did",
        "Quantum Computing's 2026 Reckoning: What's Real vs. Hype",
    ],
}

def get_gt_indices(query):
    """Return corpus indices of manually curated relevant documents for a query."""
    titles = set(MANUAL_GT.get(query, []))
    return [i for i, t in enumerate(df["title"]) if t in titles]

import re
from spellchecker import SpellChecker

# Self-contained spell checker — no dependency on query_processor or any other file
_eval_spell = SpellChecker()
with open('technical_terms.txt', 'r', encoding='utf-8') as f:
    technical_terms = [line.strip() for line in f if line.strip()]

_eval_spell.word_frequency.load_words(technical_terms)


def correct_query(query):
    words = query.strip().lower().split()
    corrected = []
    for word in words:
        clean = re.sub(r"[^\w]", "", word)
        if len(clean) <= 2 or clean.isdigit():
            corrected.append(word)
        elif clean in _eval_spell:
            corrected.append(word)
        else:
            suggestion = _eval_spell.correction(clean)
            if suggestion and suggestion != clean:
                corrected.append(word.replace(clean, suggestion))
            else:
                corrected.append(word)
    return " ".join(corrected)

TEST_QUERIES = {
    "Keyword": [
        {"query": "blockchain consensus algorithms",                          "topics": ["Blockchain"]},
        {"query": "quantum entanglement computing",                           "topics": ["Quantum Computing"]},
        {"query": "zero trust security architecture",                         "topics": ["Cybersecurity"]},
        {"query": "autonomous robot navigation SLAM",                         "topics": ["Robotics"]},
        {"query": "NLP sentiment analysis",                                   "topics": ["NLP"]},
    ],
    "Natural Language": [
        {"query": "How do neural networks learn using backpropagation",        "topics": ["AI", "Data Science", "NLP"]},
        {"query": "What are the common vulnerabilities in web applications",   "topics": ["Cybersecurity"]},
        {"query": "Explain the role of smart contracts in decentralized finance", "topics": ["Blockchain"]},
        {"query": "How does containerization work with Docker and Kubernetes", "topics": ["Cloud Computing"]},
        {"query": "What is the difference between supervised and unsupervised learning", "topics": ["AI", "Data Science"]},
    ],
    "Ambiguous": [
        {"query": "python",      "topics": ["Data Science", "NLP"]},
        {"query": "transformer", "topics": ["NLP", "AI"]},
        {"query": "cloud",       "topics": ["Cloud Computing"]},
        {"query": "security",    "topics": ["Cybersecurity"]},
        {"query": "mining",      "topics": ["Blockchain"]},
    ],
    "Noisy/Typo": [
        {"query": "machne lerning algorithams",  "topics": ["AI", "Data Science"]},
        {"query": "cybersecrity threts",          "topics": ["Cybersecurity"]},
        {"query": "nueral netwerks nlp",          "topics": ["NLP", "AI"]},
        {"query": "blockchan scalablity",         "topics": ["Blockchain"]},
        {"query": "quntum superposishon",         "topics": ["Quantum Computing"]},
    ],
}
records = []
for cat, items in TEST_QUERIES.items():
    for item in items:
        q, topics = item["query"], item["topics"]
        q_corrected = correct_query(q)   # spell-corrected form for VSM / BM25
        gt = get_gt_indices(q)
        for model_name, fn, q_input in [
            ("VSM",       vsm_retrieve,       q_corrected),
            ("BM25",      bm25_retrieve,      q_corrected),
            # Embedding receives the RAW query so we honestly measure its
            # robustness to typos — it handles semantic meaning directly
            # and does not rely on exact token matching.
            ("Embedding", embedding_retrieve, q),
        ]:
            ret = fn(q_input, 50)
            records.append({
                "Category": cat, "Query": q[:52],
                "Topics": ", ".join(topics), "GT Size": len(gt),
                "Model": model_name,
                "P@10":   round(p_at_k(ret, gt), 3),
                "Recall": round(recall_sc(ret, gt), 3),
                "AP":     round(avg_prec(ret, gt), 3),
            })

res = pd.DataFrame(records)
print(f"Evaluation complete — {len(res)} records")
res
```

We document one concrete, diagnostic failure case per model. Each failure case is run live with document-level output so the retrieval errors are directly observable.


```python
#  FAILURE CASE 1 — VSM 
# 'zero trust security architecture'
q = "zero trust security architecture"
gt = get_gt_indices(q)
top10 = vsm_retrieve(q)[:10]

print(f"VSM | Query: '{q}'")
print(f"P@10 = {p_at_k(top10, gt):.2f}  |  GT size = {len(gt)}")
print()
for rank, idx in enumerate(top10, 1):
    mark = "RELEVANT  " if idx in set(gt) else "off-topic "
    print(f"  #{rank:02d} [{mark}] [{df.iloc[idx]['topic']:16}] {df.iloc[idx]['title'][:65]}")

print()
print("ROOT CAUSE:")
print("  'security' and 'architecture' are high document-frequency tokens across all 8 topics.")
print("  Their IDF is low, meaning the VSM query vector is dominated by 'trust' — which")
print("  has moderate IDF but matches Blockchain documents about trustless systems")
print("  (e.g., 'A trustless society? A political look at the blockchain vision').")
print("  VSM treats Blockchain trust documents and Zero-Trust Cybersecurity documents as")
print("  equally relevant because they share the same token, with no semantic distinction.")
```

```python
#  FAILURE CASE 2 — BM25 
# 'How does containerization work with Docker and Kubernetes'
q = "How does containerization work with Docker and Kubernetes"
gt = get_gt_indices(q)
top10 = bm25_retrieve(q)[:10]

print(f"BM25 | Query: '{q}'")
print(f"P@10 = {p_at_k(top10, gt):.2f}  |  GT size = {len(gt)}")
print()
for rank, idx in enumerate(top10, 1):
    mark = "RELEVANT  " if idx in set(gt) else "off-topic "
    print(f"  #{rank:02d} [{mark}] [{df.iloc[idx]['topic']:16}] {df.iloc[idx]['title'][:65]}")

print()
print("ROOT CAUSE:")
print("  Only one corpus document — the AI-generated 'How containerization works with Docker")
print("  and Kubernetes' — covers both technologies together as its primary subject.")
print("  BM25 exhausts the directly relevant document in rank #1 then must fill 9 more slots.")
print("  It promotes generic Cloud Computing documents (AWS guides, Azure DevOps, Terraform)")
print("  that mention containers in passing, because BM25 has no mechanism to recognise that")
print("  these documents discuss containerization only incidentally.")
print("  This is a structural failure: when |Relevant| << k, P@10 < 1.0 is unavoidable.")
```
